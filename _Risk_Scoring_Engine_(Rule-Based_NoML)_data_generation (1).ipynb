```python
import pandas as pd
import numpy as np
from faker import Faker
import random
```

```python
# Initialize Faker and random seed
fake = Faker()
random.seed(42)
```

```python
# Parameters
n_users = 1000
min_txns = 50
max_txns = 150
```

```python
# Generateing user_ids
user_ids = [f"user_{i+1}" for i in range(n_users)]

# Create transaction data
transactions = []

for user in user_ids:
    balance = random.randint(1000, 100000)
    n_txns = random.randint(min_txns, max_txns)
    
    for _ in range(n_txns):
        txn_amount = round(random.uniform(50, 10000), 2)
        txn_type = random.choice(['debit', 'credit'])
        if txn_type == 'debit':
            balance = max(balance - txn_amount, 0)
        else:
            balance += txn_amount
        
        transactions.append({
            'user_id': user,
            'amount': txn_amount,
            'balance': balance,
            'timestamp': fake.date_time_between(start_date='-30d', end_date='now'),
            'txn_type': txn_type,
            'merchant': fake.company(),
            'location': fake.city(),
            'channel': random.choice(['online', 'POS', 'ATM']),
            'device_id': f"device_{random.randint(1, 1000)}",
            'txn_status': random.choices(['success', 'fail', 'reversed'], weights=[80, 15, 5])[0]
        })

# Createing  DataFrame

df = pd.DataFrame(transactions)
df['timestamp'] = pd.to_datetime(df['timestamp'])
df.sort_values(by=['user_id', 'timestamp'], inplace=True)
df.reset_index(drop=True, inplace=True)

# Preview
print(df.head())
```

```python
df.to_csv("sample_transaction_data.csv", index=False)
```
